# 1. Explain what Word2Vec is.
### Word2Vec is a neural network-based word embedding model that converts words into dense vector representations. It learns the meaning of words by analyzing the contexts in which they appear, so words with similar meanings have similar vector representations.

# 2. Explain the idea of predicting words from context.
### Word2Vec learns by predicting words based on their surrounding words (or predicting surrounding words from a target word). During training, it identifies patterns in word usage, allowing it to learn semantic and syntactic relationships between words.

# 3. Definitions
## • Vocabulary:
### The vocabulary is the complete set of unique words present in a text corpus that the model learns from.

## • Context Window:
### A context window is the number of neighboring words around a target word that the model considers during training. For example, a window size of 2 means the model looks at two words before and two words after the target word.

## • Embedding Dimension:
### The embedding dimension is the size (length) of the vector used to represent each word. For example, an embedding dimension of 100 represents each word as a vector containing 100 numerical values.

## 1. CBOW (Continuous Bag of Words)
## - Predicts target word from context words
## - faster and good for large datasets
## 2. Skip-Gram
## - Predicts context words from target word

## - Write 3-4 lines explaining when to use each

# 1. CBOW (Continuous Bag of Words)
## Definition:
### CBOW predicts the target (center) word using the surrounding context words.
### Example
### Sentence: "The cat sat on the mat."
### If the target word is "sat" and the context window is 2:
### Context words: The, cat, on, the
### 
### - Target word: sat
### Input: The, cat, on, the → Output: sat
### Features
### - Predicts the target word from surrounding context words.
### - Faster to train than Skip-Gram.
### - Works well with large datasets.
### - Performs better for frequent words.

# 2. Skip-Gram
## Definition:
### Skip-Gram predicts the surrounding context words from the target (center) word.
### Example
### Sentence: "The cat sat on the mat."

### If the target word is "sat":
###  Input: sat
### - Output: cat, on, The, the
### Features
### - Predicts context words from the target word.
### - Slower to train than CBOW.
### - Produces high-quality word embeddings.

# Task 4: Neural Network Intution Behind Word2vec 
# Explain:
## 1. Input layer representation.
## 2. Hidden layer (embedding layer)
## 3. Output layer
## 4. How weights become word embeddings

# 1. Input Layer Representation
### The input layer represents each word using One-Hot Encoding. If the vocabulary contains V words, the input is a vector of size V, where only one position is 1 (representing the target word) and all other positions are 0.

## Example:
## Vocabulary = [cat, dog, bird, fish]

## Word = dog

## One-Hot Vector = [0, 1, 0, 0]

# 2. Hidden Layer (Embedding Layer)
### The hidden layer is called the embedding layer. It converts the high-dimensional one-hot vector into a dense, low-dimensional vector (embedding). This layer contains the learned semantic representation of each word.

### Example:
### One-Hot: [0, 1, 0, 0]

### Embedding: [0.25, -0.61, 0.80, 0.14]

### Similar words have similar embedding vectors.

# 3. Output Layer
### The output layer predicts the target word (CBOW) or the context words (Skip-Gram). It calculates the probability of every word in the vocabulary using the Softmax function, and the word with the highest probability is selected as the prediction.

# 4. How Weights Become Word Embeddings
### Initially, the weights between the input layer and the hidden (embedding) layer are random. During training, Word2Vec updates these weights using backpropagation to minimize prediction errors. After training is complete, these learned weight vectors become the word embeddings, where words with similar meanings have similar vector representations.

# Task 5: Prepare text for word2vec
## 1. Tokenize cleaned text into sentences.
## 2. Tokenize sentences into words.
## 3. Store them as a list of lists.
## Example:
## [['this', 'movie','is', 'good'],['i','liked','this','film']]

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

# Download required resources (run once)
nltk.download('punkt')

# Sample cleaned text
text = """
This movie is good. I liked this film.
It has great acting and an interesting story.
"""

# 1. Tokenize text into sentences
sentences = sent_tokenize(text)

# 2. Tokenize each sentence into words
word_tokens = [word_tokenize(sentence.lower()) for sentence in sentences]

# 3. Store as a list of lists
print(word_tokens)
```

# Task 6: Train Word2vec Model
## Using gensim:
## 1. Train a word2vec model with:
## . vector_size=100
## . window=5
## .min_count=1
## . sg=0(CBOW)

## 2. Print:
## . Vocabulary size
## . Embedding vector for a sample word

```python


from gensim.models import Word2Vec

# Sample tokenized sentences (list of lists)
sentences = [
    ['this', 'movie', 'is', 'good'],
    ['i', 'liked', 'this', 'film'],
    ['this', 'film', 'is', 'amazing'],
    ['the', 'acting', 'was', 'excellent'],
    ['the', 'story', 'was', 'interesting']
]

# 1. Train Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0          # CBOW
)

# 2. Print vocabulary size
print("Vocabulary Size:", len(model.wv))

# Print all words in vocabulary
print("Vocabulary:", list(model.wv.index_to_key))

# Print embedding vector for a sample word
sample_word = "movie"

print(f"\nEmbedding Vector for '{sample_word}':")
print(model.wv[sample_word])

# Vector dimension
print("\nVector Shape:", model.wv[sample_word].shape)
```

# Task 7: Skip - Gram Word2Vec Model
## 1. Train another Word@Vec Model using:
## .sg = 1(Skip-Gram)
## 2. Compare:
## .Training time
## .Similar words output

```python
from gensim.models import Word2Vec
import time
```

```python
## Cbow model
start_time = time.time()

cbow_model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0         
)

# Calculate training time
cbow_time = time.time() - start_time

print("Cbow Training Time:", cbow_time)
```

```python
# Start training time
start_time = time.time()

# Train Skip-Gram Model
skipgram_model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=1,
    epochs=10
)

# Calculate training time
skipgram_time = time.time() - start_time

print("Skip-Gram Training Time:", skipgram_time)
```

```python
similar_words = skipgram_model.wv.most_similar(
    'movie',
    topn=5
)

print(similar_words)
```

```python
## Training time comaprision
print("CBOW Training Time:", cbow_time)
print("Skip-Gram Training Time:", skipgram_time)
```

```python
## Similar words outputs
print("CBOW Similar Words:")
print(cbow_model.wv.most_similar('movie', topn=5))

print("\nSkip-Gram Similar Words:")
print(skipgram_model.wv.most_similar('movie', topn=5))
```

# 1. Difference between CBOW & Skip-Gram in Practice
### . CBOW (Continuous Bag of Words): Predicts a target word using surrounding context words.
### . Skip-Gram: Predicts surrounding context words using a target word.
### . CBOW is generally faster to train and works well with frequent words.
### . Skip-Gram is generally slower but often performs better with rare words.

# 2. Advantages of Word2Vec over TF-IDF
### . Word2Vec captures semantic relationships between words.
### . It generates dense, low-dimensional word vectors.
### . It can identify similar words based on their context.
### . TF-IDF represents word importance but does not inherently understand semantic similarity.
### . Word2Vec is useful for NLP tasks such as recommendation systems and text similarity.

# 3. Limitations of Word2Vec
### . It generates a single fixed vector for each word.
### . It cannot effectively handle multiple meanings of the same word.
### . It struggles with unseen words (Out-of-Vocabulary words).
### . It requires a sufficient amount of training data.
### . It does not understand word order effectively.

# 4. Why Context Still Matters in Modern NLP
### . Context helps understand the actual meaning of words.
### .The same word can have different meanings in different sentences.
### . Modern NLP models like BERT and GPT use contextual information.
### .Context improves text understanding, sentiment analysis and language generation.
### .It helps models understand relationships between words and sentences.
